```python
# -- Cora Dataset --
import time
import random
from torch_geometric.datasets import Planetoid
import torch_geometric.utils as utils
import networkx as nx
from sklearn.metrics import roc_auc_score, average_precision_score

dataset = Planetoid(root='./data/Cora', name="Cora")
data = dataset[0]
test_ratio = 0.1

# 전체 그래프 networkx graph
g_nx = utils.to_networkx(data, to_undirected=True)
all_edges = list(g_nx.edges())
num_test_edges = int(len(all_edges) * test_ratio) # 


# 테스트용 양성 샘플 선택 후 그래프에서 제거 (학습용 그래프 생성)
random.seed(42)
test_edges_pos = random.sample(all_edges, num_test_edges)
train_g = g_nx.copy()
train_g.remove_edges_from(test_edges_pos)

# 음성 샘플 샘플링(=pos)
test_edges_neg = []
nodes = list(g_nx.nodes())
while len(test_edges_neg) < num_test_edges:
    u, v = random.sample(nodes, 2)
    if u != v and not g_nx.has_edge(u, v):
        test_edges_neg.append((u, v))

def get_aa_score(graph, edge_list):
    preds = nx.adamic_adar_index(graph, edge_list)
    return [score for u, v, score in preds]

start_time = time.time()
pos_scores = get_aa_score(train_g, test_edges_pos)
neg_scores = get_aa_score(train_g, test_edges_neg)

y_true = [1] * len(pos_scores) + [0] * len(neg_scores)
y_scores = pos_scores + neg_scores

auc = roc_auc_score(y_true, y_scores)
ap = average_precision_score(y_true, y_scores)
print(f'--test pos edge number: {num_test_edges}, test neg edge number: {len(test_edges_neg)}')
print(f"Planetoid Cora 데이터셋 기준 Adamic-Adar 링크 예측 AUC: {auc:.4f}, AP: {ap:.4f}")
print(f"걸린 시간: {time.time() - start_time:.4f} (s)")

```

```python
# -- davis_southern_women Dataset: 2-mode --
import math
import random
import time
import networkx as nx
from sklearn.metrics import roc_auc_score, average_precision_score

# 1. Davis Southern Women (2-mode) 데이터셋 로드
G = nx.davis_southern_women_graph()

# Node Set U (여성)와 Node Set V (이벤트) 분리
women = [n for n, d in G.nodes(data=True) if d['bipartite'] == 0]
events = [n for n, d in G.nodes(data=True) if d['bipartite'] == 1]

# 전체 엣지 및 테스트 비율 설정
all_edges = list(G.edges())
test_ratio = 0.2  # 소규모 그래프이므로 테스트 비율을 20%로 설정
num_test_edges = int(len(all_edges) * test_ratio)

# 2. 테스트용 양성 샘플(Pos) 선택 및 학습용 그래프(Train Graph) 분리
random.seed(42)
test_edges_pos = random.sample(all_edges, num_test_edges)

train_g = G.copy()
train_g.remove_edges_from(test_edges_pos)

# 3. 음성 샘플(Neg) 샘플링 (2-mode 조건 준수: u in U, v in V 간 미연결 쌍만 추출)
test_edges_neg = []
while len(test_edges_neg) < num_test_edges:
    u = random.choice(women)
    v = random.choice(events)
    # 실제 그래프 G에 연결이 없고, 음성 샘플에 중복 추가되지 않은 경우
    if not G.has_edge(u, v) and (u, v) not in test_edges_neg:
        test_edges_neg.append((u, v))


# 4. 2-mode 전용 Adamic-Adar Score 계산 함수 정의
def bipartite_adamic_adar_score(graph, u, v):
    """
    2-mode 그래프에서 노드 u(Set U)와 노드 v(Set V) 간의 AA 예측 점수 산출
    """
    score = 0.0
    # v와 연결된 다른 노드들 u' (Set U)
    for u_prime in graph.neighbors(v):
        if u_prime == u:
            continue
        
        # u와 u'이 공유하는 공통 이웃 노드들 v_common (Set V)
        common_neighbors = set(graph.neighbors(u)).intersection(set(graph.neighbors(u_prime)))
        
        for v_common in common_neighbors:
            deg = graph.degree(v_common)
            if deg > 1:
                score += 1.0 / math.log(deg)
                
    return score


def get_bipartite_aa_scores(graph, edge_list):
    scores = []
    for u, v in edge_list:
        # 엣지 튜플의 노드 순서가 (v, u)로 들어오는 경우를 대비한 노드 타입 정렬
        if u in events and v in women:
            u, v = v, u
        scores.append(bipartite_adamic_adar_score(graph, u, v))
    return scores


# 5. 링크 예측 수행 및 성능 평가
start_time = time.time()

pos_scores = get_bipartite_aa_scores(train_g, test_edges_pos)
neg_scores = get_bipartite_aa_scores(train_g, test_edges_neg)

y_true = [1] * len(pos_scores) + [0] * len(neg_scores)
y_scores = pos_scores + neg_scores

auc = roc_auc_score(y_true, y_scores)
ap = average_precision_score(y_true, y_scores)

# 결과 출력
print(f"-- test pos edge number: {num_test_edges}, test neg edge number: {len(test_edges_neg)}")
print(f"Davis Southern Women 데이터셋 기준 Bipartite Adamic-Adar 링크 예측 AUC: {auc:.4f}, AP: {ap:.4f}")
print(f"걸린 시간: {time.time() - start_time:.4f} (s)")
```

```python
import math
import random
import time
import torch
import networkx as nx
from torch_geometric.datasets import MovieLens
from sklearn.metrics import roc_auc_score, average_precision_score

# 1. MovieLens 데이터셋 로드 (MovieLens-100k)
dataset = MovieLens(root='./data/MovieLens')
data = dataset[0]

# PyG HeteroData 구조에서 User-Movie 연결 정보 추출
# edge_index: [2, num_edges] (row 0: user_id, row 1: movie_id)
user_src = data['user', 'rates', 'movie'].edge_index[0].tolist()
movie_dst = data['user', 'rates', 'movie'].edge_index[1].tolist()

# Bipartite 그래프 생성을 위해 Movie ID에 오프셋(Prefix) 적용 (노드 ID 충돌 방지)
user_nodes = list(set(user_src))
movie_nodes = [f"m_{m}" for m in set(movie_dst)]

G = nx.Graph()
G.add_nodes_from(user_nodes, bipartite=0)
G.add_nodes_from(movie_nodes, bipartite=1)

# 엣지 추가
all_edges = [(u, f"m_{m}") for u, m in zip(user_src, movie_dst)]
G.add_edges_from(all_edges)

# 테스트 비율 설정 (10%)
test_ratio = 0.1
num_test_edges = int(len(all_edges) * test_ratio)

# 2. Train / Test (Pos) Split
random.seed(42)
test_edges_pos = random.sample(all_edges, num_test_edges)

train_g = G.copy()
train_g.remove_edges_from(test_edges_pos)

# 3. 음성 샘플(Neg) 샘플링 (u in User, v in Movie 간 미연결 쌍)
test_edges_neg = []
all_edges_set = set(all_edges)
while len(test_edges_neg) < num_test_edges:
    u = random.choice(user_nodes)
    v = random.choice(movie_nodes)
    if (u, v) not in all_edges_set and (u, v) not in test_edges_neg:
        test_edges_neg.append((u, v))


# 4. 2-mode 휴리스틱 계산 함수 정의

# ① Bipartite Adamic-Adar
def bipartite_adamic_adar_score(graph, u, v):
    score = 0.0
    for u_prime in graph.neighbors(v):
        if u_prime == u:
            continue
        common_neighbors = set(graph.neighbors(u)).intersection(set(graph.neighbors(u_prime)))
        for v_common in common_neighbors:
            deg = graph.degree(v_common)
            if deg > 1:
                score += 1.0 / math.log(deg)
    return score

# ② Bipartite Jaccard
def bipartite_jaccard_score(graph, u, v):
    score = 0.0
    u_neighbors = set(graph.neighbors(u))
    if not u_neighbors:
        return 0.0
    for u_prime in graph.neighbors(v):
        if u_prime == u:
            continue
        u_prime_neighbors = set(graph.neighbors(u_prime))
        intersection = len(u_neighbors.intersection(u_prime_neighbors))
        union = len(u_neighbors.union(u_prime_neighbors))
        if union > 0:
            score += intersection / union
    return score

# Score 일괄 추출 스크립트
def get_scores(graph, edge_list, method='adamic_adar'):
    scores = []
    for u, v in edge_list:
        # 노드 순서 정렬 (u: user, v: movie)
        if str(u).startswith('m_'):
            u, v = v, u
            
        if method == 'adamic_adar':
            scores.append(bipartite_adamic_adar_score(graph, u, v))
        elif method == 'jaccard':
            scores.append(bipartite_jaccard_score(graph, u, v))
    return scores


# 5. 링크 예측 수행 및 성능 평가
method_type = 'adamic_adar'  # 'adamic_adar' 또는 'jaccard' 선택

start_time = time.time()
pos_scores = get_scores(train_g, test_edges_pos, method=method_type)
neg_scores = get_scores(train_g, test_edges_neg, method=method_type)

y_true = [1] * len(pos_scores) + [0] * len(neg_scores)
y_scores = pos_scores + neg_scores

auc = roc_auc_score(y_true, y_scores)
ap = average_precision_score(y_true, y_scores)

# 결과 출력
print(f"-- test pos edge number: {num_test_edges}, test neg edge number: {len(test_edges_neg)}")
print(f"MovieLens-100k 데이터셋 기준 Bipartite {method_type.upper()} 링크 예측 AUC: {auc:.4f}, AP: {ap:.4f}")
print(f"걸린 시간: {time.time() - start_time:.4f} (s)")
```
